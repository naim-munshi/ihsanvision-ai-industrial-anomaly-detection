```python
import os
import torch
import anomalib

print("PyTorch:", torch.__version__)
print("Anomalib:", anomalib.__version__)
```

```python
DATASET_PATH = "/kaggle/input/datasets/ipythonx/mvtec-ad"

OUTPUT_PATH = "/kaggle/working/ihsanvision_outputs"

os.makedirs(OUTPUT_PATH, exist_ok=True)

print(DATASET_PATH)
print(OUTPUT_PATH)
```

```python
categories = os.listdir(DATASET_PATH)

print("Categories:")
for category in categories:
    print(category)
```

```python
CATEGORY = "bottle"

print(
    "Training category:",
    CATEGORY
)
```

```python
from anomalib.data import MVTecAD

datamodule = MVTecAD(
    root=DATASET_PATH,
    category=CATEGORY
)

datamodule.setup()

print("Dataset loaded successfully")
```

```python
from anomalib.models import Padim

model = Padim()

print("PaDiM model created")
```

```python
from anomalib.engine import Engine

engine = Engine(
    default_root_dir=OUTPUT_PATH
)

print("Engine ready")
```

```python
engine.fit(
    model=model,
    datamodule=datamodule
)
```

```python
engine.test(
    model=model,
    datamodule=datamodule
)
```
